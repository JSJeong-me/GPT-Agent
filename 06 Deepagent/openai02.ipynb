```python
import os
from getpass import getpass

# 'OPENAI_API_KEY'를 안전하게 입력받아 환경 변수에 설정합니다.
api_key = getpass("OpenAI API Key를 입력하세요: ")
os.environ["OPENAI_API_KEY"] = api_key
print("OPENAI_API_KEY가 환경 변수로 설정되었습니다.")
```

```python
from openai import OpenAI
client = OpenAI()

stream = client.responses.create(
  model="gpt-6-luna",
  service_tier="default",
  input=[
    {
      "role": "system",
      "content": [
        {
          "type": "input_text",
          "text": "당신은 소설가이고, 나는 당신의 주인입니다."
        }
      ]
    },
    {
      "role": "user",
      "content": [
        {
          "type": "input_text",
          "text": "인터넷 검색을 사용하여 '부캉이'에 대해 검색해보고, 검색 결과를 바탕으로 부캉이를 주제로 재미난 이야기 3줄 작성해 주세요."
        }
      ]
    }
  ],
  text={
    "format": {
      "type": "text"
    }
  },
  reasoning={},
  tools=[
    {
      "type": "web_search_preview"
    }
  ],
  temperature=1,
  max_output_tokens=2048,
  stream=True,
  store=True,
  include=["web_search_call.action.sources"]
)

for event in stream:
    if event.type in ("response.output_text.delta", "response.refusal.delta"):
        print(event.delta, end="", flush=True)
    elif event.type == "error":
        raise RuntimeError(event.message)
    elif event.type == "response.failed":
        raise RuntimeError(event.response.error)
```

```python
# 결과를 저장할 변수 초기화 (곱셈이므로 1로 설정)
result = 1

# 1부터 10까지 반복
for i in range(1, 11):
    result *= i

print(f"1부터 10까지의 곱: {result}")
```

```python
def fibonacci(n):
    if n <= 0:
        return "Input should be a positive integer"
    elif n == 1:
        return 0
    elif n == 2:
        return 1

    a, b = 0, 1
    for _ in range(2, n):
        a, b = b, a + b
    return b

# 예시: 첫 10개의 피보나치 수열을 계산
for i in range(1, 11):
    print(f"{i}: {fibonacci(i)}")
```
